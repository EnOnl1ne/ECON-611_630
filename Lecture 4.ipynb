```python
import numpy as np
a=np.array([0.1, 1.2, 2.3, 3.4, 9])
print(a)
```

```python
#initialize a numpy array as a pandas series
s=pd.Series(a)
```

```python
#Series indexes don't have to be integers
s = pd.Series(a, index=['a', 'b', 'c', 'd', 'e'])
s
```

```python
#Select element at 0 location
s.iloc[0]
```

```python
#Select element with index a
s.loc['a']
```

```python
#Select all elements from 2 to 4
s.iloc[2:4]
```

```python
#Unlike numpy, indexes need not be unique
s=pd.Series([0.1, 1.2, 2.3, 3.4, 9], index = ['a', 'b', 'c', 'a', 'a'])
s['a']
```

```python
#Often build panda series with dictionaries
s1 = pd.Series({'a': 0.1, 'b': 1.2, 'c': 2.3})
```

```python
state_gdp = pd.read_excel('US_state_GDP.xls', 'Sheet1')
```

```python
state_gdp.head()
```

```python
#Select by column
state_gdp['gdp_2009'].head()
```

```python
#Drop = True -> state set as an index, dropped out of the data so it doesn't become a self-reference
state_gdp = state_gdp.set_index('state', drop=True)
state_gdp.head()
```

```python
state_gdp.loc['Alaska', ['state_code', 'region']]
```

```python
#Or...
state_gdp.iloc[0,0]
```

```python
#Filtering data: show only data for which gdp in 2012 was higher than median gdp in the same year
state_gdp[state_gdp['gdp_2012']>state_gdp['gdp_2012'].median()].head()
```

```python
new_df = state_gdp[state_gdp['gdp_2012']>state_gdp['gdp_2012'].median()].head()
new_df
```

```python
#group by region and mean gdp_2012 for each region
state_gdp.groupby('region').agg({'gdp_2012': ['mean', 'std']})
```
