### Problem 3

```python
y = np.array([0,0.5,1.5,2.5,1.0,1.0001,-0.5,-1,-1.5,-2.5])
y
```

```python
y.round(2)
```

```python
y.round()
```

```python
#Or, alternatively...
np.around(y, 2)
```

```python
np.around(y)
```

```python
np.floor(y)
```

```python
np.ceil(y)
```

In words, round() and around() will round the values in the array to the nearest integer, unless we specify decimal place precision, in which case it rounds figures to that many decimal places. This means that values <= 0.5 are rounded down to zero, and everything 0.5 < x <=1 rounds up to 1. floor() will round down to the nearest lower integer, so x<1 rounds down to 0. ceil() rounds up to the nearest larger integer, so x<=1 rounds up to 1.
