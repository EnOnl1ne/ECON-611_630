## 0. Setup

You already downloaded both of these from FRED in week 2. Put them in the same folder as this notebook.

| Series | What it is | Units |
|---|---|---|
| UNRATE | Unemployment rate | percent, monthly, 1948 to today |
| USREC | NBER recession indicator | 1 = recession month, 0 = not |

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

%matplotlib inline
```

Two files, two `read_csv` calls, the same three arguments each time (week 2, Sheppard ch. 8):

- `parse_dates` turns the date column into real dates,
- `index_col` makes that column the index, so we can slice by date,
- `['SERIES']` at the end takes the one data column out of the DataFrame, which leaves a Series.

```python
unrate = pd.read_csv('UNRATE.csv', parse_dates=['observation_date'],
                     index_col='observation_date')['UNRATE']

usrec = pd.read_csv('USREC.csv', parse_dates=['observation_date'],
                    index_col='observation_date')['USREC']

unrate.tail()
```

A level and a change answer different questions, so we build the change once and keep it.

`unrate.shift(12)` slides the column down twelve rows, so each month sits next to the same month a year earlier. Subtracting gives the change from a year ago, in percentage points. It is the labelled version of `x[1:] - x[:-1]` from week 2: the index does the lining up for you. (`unrate.diff(12)` is the shorthand for the same thing.)

Then we put the series side by side in one DataFrame, built from a dictionary of Series exactly as in week 2. Because they share a date index, pandas lines them up by date: `USREC` starts in 1854 and `UNRATE` in 1948, and we keep the overlap.

```python
chg12 = unrate - unrate.shift(12)

df = pd.DataFrame({'UNRATE': unrate, 'CHG12': chg12, 'USREC': usrec})
df = df.loc['1948':]
df.tail()
```

```python
# look before you leap: how many rows, and what is missing?
print(df.shape)
df.isnull().sum()
```

## 1. Figure and Axes

Matplotlib has two interfaces.

- **pyplot style**: call `plt.plot`, `plt.title`, ... and matplotlib keeps track of the current figure for you. Short, and fine for a quick look. This is what Sheppard's chapter uses and what you used in HW4.
- **Object style**: ask for the objects explicitly with `fig, ax = plt.subplots()`, then call methods on them.

A **Figure** is the whole image. An **Axes** is one plotting area inside it, with its own x-axis, y-axis, title and legend. A figure can hold several axes.

### Building the x values

`np.linspace(start, stop, n)` returns `n` evenly spaced points from `start` to `stop`, with **both endpoints included**. So `np.linspace(1, 10, 200)` is 200 points from 1 to 10, about 0.045 apart.

This is new. In Lecture 9 we built arrays with `np.arange(start, stop, step)`, where you give the **step** and the stop value is left out. `linspace` flips that: you give the **count** and get both ends. For a plotting grid the count is what you care about, since it controls how smooth the curve looks. 200 points draws a clean sine wave; 20 would look jagged.

Both work. `linspace` is shorter and doesn't make you worry about whether the endpoint sneaks in or gets dropped, which is why it's the standard way to set up a plotting grid.

A **format string** sets color, marker and line style in one short code (Sheppard's table):

| Color | | Marker | | Line | |
|---|---|---|---|---|---|
| `b` blue | `k` black | `o` circle | `s` square | `-` solid | `--` dashed |
| `g` green | `r` red | `^` triangle | `x` cross | `-.` dash-dot | `:` dotted |

`'ko--'` is a black dashed line with circles. Keyword arguments give finer control: `color`, `linestyle`, `linewidth`, `marker`, `markersize`, `alpha` (transparency) and `label` (used by the legend).

```python
x = np.linspace(1, 10, 200)
y = np.sin(x)

# pyplot style
#plt.plot(x, y, 'b-', linewidth=2)
plt.plot(x, y, 'b-', linewidth=2)
plt.show()
```

```python
# object style: same picture
fig, ax = plt.subplots()
ax.plot(x, y, 'b-', linewidth=2)
plt.show()
```

```python
# fig and ax are ordinary Python objects
type(fig), type(ax)
```

We use the object style for the rest of the course. With more than one panel it is the only way to say *which* panel you mean.

Most pyplot functions have an `ax.set_...` twin:

| pyplot | object style |
|---|---|
| `plt.title('...')` | `ax.set_title('...')` |
| `plt.xlabel('...')` | `ax.set_xlabel('...')` |
| `plt.ylim(0, 16)` | `ax.set_ylim(0, 16)` |
| `plt.xticks([...])` | `ax.set_xticks([...])` |
| `plt.legend()` | `ax.legend()` |
| `plt.savefig('f.png')` | `fig.savefig('f.png')` |

## 3. Scatter plots

A scatter plot answers *do these two variables move together?* (Sheppard 15.2.3).

We only have one series, so the second variable comes from it: the level today against the change over the past year. Unemployment that is high *and* rising is a different situation from unemployment that is high and falling, and a scatter separates them.

```python
fig, ax = plt.subplots()
ax.scatter(x=df['UNRATE'], y=df['CHG12'], s=10, alpha=.4)
ax.axhline(0, color='grey', linewidth=0.8)
ax.set_xlabel('Unemployment rate (%)')
ax.set_ylabel('Change over the past 12 months (pp)')
plt.show()
```

**Question:** The cloud leans. What does that lean say about how unemployment behaves when it is already high? And what is this chart hiding? (Each dot is a month. Which dots are next to each other in time?)

`s` sets marker size, `c` sets color, `alpha` sets transparency. Sheppard's trick: pass an *array* to `s` or `c` and the chart carries a third variable. Here color shows the year.

```python
fig, ax = plt.subplots(figsize=(7, 5))
points = ax.scatter(df['UNRATE'], df['CHG12'], c=df.index.year, s=10, cmap='viridis')
fig.colorbar(points, ax=ax, label='Year')
ax.axhline(0, color='grey', linewidth=0.8)
ax.set_xlabel('Unemployment rate (%)')
ax.set_ylabel('Change over the past 12 months (pp)')
plt.show()
```

A colorbar is hard to read precisely. When the third variable has only a few values, split the data with a boolean mask (week 2), give each group its own `scatter` call and `label`, then add a legend. `USREC` splits the months into recession and expansion.

```python
rec = df[df['USREC'] == 1]
exp = df[df['USREC'] == 0]

fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(exp['UNRATE'], exp['CHG12'], s=12, alpha=0.5,
           color='#9AAAB1', label='Expansion month')
ax.scatter(rec['UNRATE'], rec['CHG12'], s=12, alpha=0.8,
           color='#C4602A', label='Recession month')
ax.axhline(0, color='grey', linewidth=0.8)
ax.set_xlabel('Unemployment rate (%)')
ax.set_ylabel('Change over the past 12 months (pp)')
ax.legend()
plt.show()
```

**Question:** Almost every orange dot sits above the zero line, but plenty of grey dots sit there too. If you had to call a recession from this chart alone, what rule would you use, and how often would it be wrong? (This is roughly the idea behind the Sahm rule.)

## 4. Bar charts, histograms, and a pie chart

- **Bar chart**: compare a number across categories (Sheppard 15.2.4).
- **Histogram**: show how one variable is distributed (Sheppard 15.2.6).
- **Pie chart**: show parts of a whole (Sheppard 15.2.5). Economists rarely use them; we will see why.

Average unemployment by decade is a groupby from last week. The bar chart is one more line.

```python
since1950 = df.loc['1950':].copy()

# a column to group on: 1953 -> 1950, 1967 -> 1960, ...
since1950['decade'] = since1950.index.year // 10 * 10

avg_unrate = since1950.groupby('decade')['UNRATE'].mean()
avg_unrate
```

```python
labels = [str(d) + 's' for d in avg_unrate.index]

fig, ax = plt.subplots()
ax.bar(labels, avg_unrate.values, color='#2F6DB5')
ax.set_ylabel('Average unemployment rate (%)')
ax.set_title('Average unemployment by decade (2020s to date)')
plt.show()
```

**Question:** A bar's *length* is what the eye reads. What happens to this chart if the y-axis starts at 4 instead of 0? Try `ax.set_ylim(4, 7.5)`. When is it fine for a line chart's axis not to start at zero?

```python
# Your turn: the same numbers as a horizontal bar chart (ax.barh),
# sorted from highest to lowest, highest at the top.
# Hint: sort_values(), and ax.invert_yaxis() puts the first bar at the top

ranked = avg_unrate.sort_values(ascending=False)
fig, ax = plt.subplots()
ax.barh([str(d) + 's' for d in ranked.index], ranked.values, color='#2F6DB5')
ax.invert_yaxis()
ax.set_xlabel('Average unemployment rate (%)')
plt.show()
```

Now a histogram of the *monthly* change in the unemployment rate.

```python
change = df['UNRATE'].diff().dropna()

fig, ax = plt.subplots()
ax.hist(change, bins=100)
ax.set_xlabel('Monthly change in unemployment rate (percentage points)')
ax.set_ylabel('Number of months')
plt.show()
```

**Question:** Why does almost all the data sit in one or two bars? Find the month responsible.

```python
change.idxmax(), change.max()
```

Sheppard 15.3.2: two plots on the same axes. Draw the histogram, then draw a normal curve with the same mean and standard deviation on top. `density=True` rescales the bars so their total area is 1, which puts them on the same footing as the curve.

The formula in the next cell is the normal density from week 2, written with NumPy.

```python
m = change.mean()
sd = change.std()

grid = np.linspace(-1, 1, 200)
normal = np.exp(-0.5 * ((grid - m) / sd) ** 2) / (sd * np.sqrt(2 * np.pi))

fig, ax = plt.subplots()
ax.hist(change, bins=np.arange(-1, 1.05, 0.1), density=True,
        color='#9AAAB1', label='Data')
ax.plot(grid, normal, color='#C4602A', linewidth=2,
        label='Normal, same mean and sd')
ax.set_xlim(-1, 1)
ax.set_xlabel('Monthly change in unemployment rate (percentage points)')
ax.legend()
plt.show()
```

**Question:** The normal curve is too flat and too wide. Name two reasons. (One is a single month. The other is about how unemployment usually moves from one month to the next.)

Last, a pie chart: the share of months since 1948 the US spent in recession.

```python
counts = df['USREC'].value_counts()
counts = counts.loc[[0, 1]]        # expansion months first, then recession months

fig, ax = plt.subplots()
ax.pie(counts.values, labels=['Expansion', 'Recession'], autopct='%1.0f%%',
       colors=['#9AAAB1', '#C4602A'])
plt.show()
```

**Question:** The pie chart delivers one number: 13%. Is a chart the best way to say that? When would a pie chart be worse than a bar chart? (Try comparing five slices of similar size.)

## 5. Several lines on one axes, and log scales

Call `ax.plot` more than once and the lines share the axes. Give each a `label` and call `ax.legend()` (Sheppard 15.3.3).

```python
fig, ax = plt.subplots(figsize=(10, 4.5))
ax.plot(df['UNRATE'], label='Unemployment rate')
ax.plot(df['CHG12'], label='Change over the past 12 months')
ax.axhline(0, color='grey', linewidth=0.8)
ax.set_ylabel('Percent / percentage points')
ax.legend()
plt.show()
```

**Question:** Both lines are measured in percentage points, so nothing is technically wrong. Is the chart still misleading? What does the reader have to hold in their head to read the orange line correctly? (Section 6 has the usual fix.)

A second use for two lines: a **log scale**. Growth is about *percent* changes, and on a log scale equal percent changes look like equal steps. Two economies, one growing 2% a year and one 5%, over sixty years:

```python
years = np.arange(0, 61)
slow = 100 * 1.02 ** years
fast = 100 * 1.05 ** years

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

axes[0].plot(years, slow, label='2% a year')
axes[0].plot(years, fast, label='5% a year')
axes[0].set_title('Linear scale')
axes[0].legend()

axes[1].plot(years, slow, label='2% a year')
axes[1].plot(years, fast, label='5% a year')
axes[1].set_yscale('log')
axes[1].set_title('Log scale')
axes[1].legend()

for i in range(2):
    axes[i].set_xlabel('Years')
plt.show()
```

**Question:** On the linear chart, in which decade can you first *see* that the two economies differ? On the log chart, what does a straight line mean, and what does a steeper straight line mean? This is why GDP, prices and population are usually plotted on a log scale.

Plotting in a loop, as in QuantEcon exercise 11.1. Each call adds a line, and matplotlib picks the next color for you.

```python
# Your turn (QuantEcon exercise 11.1): on one axes, plot
#   f(x) = cos(pi * theta * x) * exp(-x)
# over x in [0, 5], for each theta in np.linspace(0, 2, 10).

def f(x, theta):
    return np.cos(np.pi * theta * x) * np.exp(-x)

x = np.linspace(0, 5, 200)
fig, ax = plt.subplots()
for theta in np.linspace(0, 2, 10):
    ax.plot(x, f(x, theta))
plt.show()
```

## 6. Subplots

`plt.subplots(nrows, ncols)` returns one figure and an *array* of axes. Pick a panel with `axes[row, col]`, the same indexing as a NumPy matrix. (Sheppard builds the same thing with `fig.add_subplot(2, 2, i)`, counting panels from 1.)

Small multiples are the honest way to compare episodes: one recession per panel, every panel drawn the same way. `sharey=True` gives every panel the same y-axis, so a height in one panel means the same as that height in another.

```python
fig, axes = plt.subplots(2, 2, figsize=(11, 7), sharey=True)

axes[0, 0].plot(df.loc['1972':'1978', 'UNRATE'])
axes[0, 0].set_title('1973-75')

axes[0, 1].plot(df.loc['1979':'1986', 'UNRATE'])
axes[0, 1].set_title('1980 and 1981-82')

axes[1, 0].plot(df.loc['2006':'2017', 'UNRATE'])
axes[1, 0].set_title('2008-09')

axes[1, 1].plot(df.loc['2019':, 'UNRATE'])
axes[1, 1].set_title('2020')

fig.tight_layout()
plt.show()
```

`fig.tight_layout()` removes wasted space and stops titles from overlapping (Sheppard 15.2.1).

Four near-identical blocks is the pattern a loop removes. `axes.flat` is the four panels in a row, so the loop from week 1 works on it: build one list per thing that changes, then walk through them by position.

```python
starts = ['1972', '1979', '2006', '2019']
ends = ['1978', '1986', '2017', '2026']
titles = ['1973-75', '1980 and 1981-82', '2008-09', '2020']

fig, axes = plt.subplots(2, 2, figsize=(11, 7), sharey=True)

for i in range(len(starts)):
    window = df.loc[starts[i]:ends[i], 'UNRATE']
    axes.flat[i].plot(window, color='#2F6DB5', linewidth=1.2)
    axes.flat[i].set_title(titles[i])

axes[0, 0].set_ylabel('Unemployment rate (%)')
axes[1, 0].set_ylabel('Unemployment rate (%)')
fig.tight_layout()
plt.show()
```

**Question:** The shared y-axis makes one comparison easy and one hard. Which is which? (Look at the 1973-75 panel, then at 2020.) If the point of your chart were the 1970s, what would you change?

## 7. Dates on the x-axis and shaded recessions

Sheppard shades NBER recessions behind capacity utilization (15.3.5). The recipe is the same for any monthly series.

`ax.fill_between(x, y1, y2, where=condition)` fills the band between `y1` and `y2` only where the condition is `True`. `USREC == 1` is exactly that condition.

```python
fig, ax = plt.subplots(figsize=(11, 4.5))
ax.plot(df.index, df['UNRATE'], color='black', linewidth=1)
ax.fill_between(df.index, 0, 16, where=df['USREC'] == 1,
                color='grey', alpha=.3, label='NBER recession')
ax.set_ylim(0, 16)
ax.set_xlim(df.index[0], df.index[-1])
ax.set_ylabel('Unemployment rate (%)')
ax.legend(loc='upper left')
plt.show()
```

`fill_between` needs `y1` and `y2` to cover the whole height you want shaded, which is why they match `set_ylim`.

**Question:** Unemployment keeps rising after the shading stops, every time. Why would the NBER call a recession over while unemployment is still going up?

Dates on an axis are handled by *locators* (where the ticks go) and *formatters* (how the tick labels read), from `matplotlib.dates` (Sheppard 15.3.4). A tick every 10 years:

```python
ax.xaxis.set_major_locator(mdates.YearLocator(10))
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
fig
```

Because we kept `fig` and `ax`, we changed the finished chart and displayed it again by typing `fig`. With the pyplot style that chart would be gone.

Zoomed in on 2020, ten-year ticks are useless. Put a tick every three months and show month and year. `fig.autofmt_xdate()` rotates the labels so they do not collide.

```python
covid = df.loc['2019-07':'2021-06', 'UNRATE']

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(covid, marker='o', markersize=3)
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=3))
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
fig.autofmt_xdate()
ax.set_ylabel('Unemployment rate (%)')
plt.show()
```

Any time the automatic tick labels are unreadable or misleading, a locator and a formatter are the fix.

Format codes: `%Y` 2026, `%y` 26, `%m` 09, `%b` Sep, `%B` September, `%d` day of the month.

## 8. From a chart for you to a chart for someone else (about 7 min)

Everything so far was **exploratory**: charts that help *you* see the data. A chart for someone else is **explanatory**. The reader has a few seconds and cannot ask you what you meant.

Before drawing, write four lines:

- **Question**: what is the reader trying to find out?
- **Answer**: the one sentence you want them to leave with. This becomes the title.
- **Evidence**: which data, which period, which chart type.
- **Reader**: who they are and what they already know.

Our example:

- **Question**: how long does it take the labor market to recover from a recession?
- **Answer**: *after 2008 unemployment took nine years to get back to where it started; after 2020 it took a little over two.*
- **Evidence**: UNRATE, monthly, 1948 to today, recessions shaded, the two return months marked.
- **Reader**: an economics student who knows what a recession is and has not looked at this series.

**Check first.** A title is a claim. Test it before you publish it.

```python
# the level each recovery had to get back to
pre_2008 = df.loc['2007', 'UNRATE'].min()
pre_2020 = df.loc['2019-09':'2020-02', 'UNRATE'].min()

# the first month after the recession started that got back there
after_2008 = df.loc['2008':, 'UNRATE']
after_2020 = df.loc['2020-03':, 'UNRATE']
back_2008 = after_2008[after_2008 <= pre_2008].index[0]
back_2020 = after_2020[after_2020 <= pre_2020].index[0]

print('2008: from', pre_2008, '% in 2007, back there in', back_2008.date())
print('2020: from', pre_2020, '% in early 2020, back there in', back_2020.date())
```

```python
months_2008 = (back_2008.year - 2008) * 12 + (back_2008.month - 1)
months_2020 = (back_2020.year - 2020) * 12 + (back_2020.month - 3)

# how deep: how far unemployment rose above where it started
peak_2008 = df.loc['2008':'2012', 'UNRATE'].max()
peak_2020 = df.loc['2020':'2021', 'UNRATE'].max()

print('2008: peak', peak_2008, '| rose', round(peak_2008 - pre_2008, 1),
      'pp | recovered in', months_2008, 'months')
print('2020: peak', peak_2020, '| rose', round(peak_2020 - pre_2020, 1),
      'pp | recovered in', months_2020, 'months')
print('2020 was', round((peak_2020 - pre_2020) / (peak_2008 - pre_2008), 1),
      'times as deep and', round(months_2008 / months_2020, 1), 'times as fast')
```

Unemployment rose 11.3 points in 2020 against 5.6 in 2008, and came back in 28 months against 110. Twice as deep, four times as fast. That contrast is the point worth making, so the title says it -- and note that "deep" in the title means the rise, not the peak.

**Question:** "Back to where it started" is our definition, not a fact. Name one other definition of "recovered" that would change the answer. Does the chart still support the title under your definition?

Now build the chart around the sentence.

```python
fig, ax = plt.subplots(figsize=(11, 5))

ax.fill_between(df.index, 0, 16, where=df['USREC'] == 1,
                color='#9AAAB1', alpha=0.35, linewidth=0, label='Recession')
ax.plot(df.index, df['UNRATE'], color='#2F6DB5', linewidth=1.3,
        label='Unemployment rate')

# mark the two months the recoveries finished
ax.scatter([back_2008, back_2020], [pre_2008, pre_2020],
           color='#C4602A', zorder=3, s=30)

ax.annotate('Mar 2017: back to 4.4%,\n110 months after 2008',
            xy=(back_2008, pre_2008), xytext=(-25, 105),
            textcoords='offset points', ha='right', fontsize=9,
            arrowprops={'arrowstyle': '->', 'color': '#C4602A'})

ax.annotate('Jul 2022: back to 3.5%,\n28 months after 2020',
            xy=(back_2020, pre_2020), xytext=(18, -55),
            textcoords='offset points', ha='left', fontsize=9,
            arrowprops={'arrowstyle': '->', 'color': '#C4602A'})

# the title states the finding; the subtitle says what is plotted
fig.suptitle('The 2020 shock was twice as deep as 2008 and recovered four times faster',
             x=0.06, ha='left', fontsize=14, fontweight='bold')
ax.set_title('US unemployment rate, percent, monthly. Shaded: NBER recessions. '
             'Marked: the first month back at the pre-recession rate.',
             loc='left', fontsize=9, color='dimgray')

ax.set_ylim(0, 16)
ax.set_xlim(df.index[0], df.index[-1])
ax.set_ylabel('Percent of the labor force')
ax.xaxis.set_major_locator(mdates.YearLocator(10))
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y'))
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.legend(loc='upper left', frameon=False)

fig.text(0.06, 0.01, 'Source: BLS and NBER via FRED (UNRATE, USREC). '
         'October 2025 not collected.', fontsize=8, color='dimgray')
fig.tight_layout(rect=(0, 0.03, 1, 1))
plt.show()
```

What changed from the chart in section 7, and why:

| Change | Why |
|---|---|
| Title states the finding | Readers read the title first and often only the title |
| Subtitle names the series, units, frequency | The title no longer has room for that |
| Two annotations | The title is about two dates, so the chart points at those two dates |
| Source note, and the missing month declared | Anyone can find and check the data |
| No top and right spines, no legend frame | Less ink that carries no information |

## 9. House style and saving figures (a 2-minute demo; the saving part you will need for the homework)

If you apply the same tweaks to every chart, put them in a function (QuantEcon 11.3.4). It calls `plt.subplots`, adjusts the axes, and hands `fig, ax` back.

```python
def econ_axes(figsize=(9, 4.5)):
    """plt.subplots() with the course's house style."""
    fig, ax = plt.subplots(figsize=figsize)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.grid(axis='y', alpha=0.3)
    return fig, ax

fig, ax = econ_axes()
ax.plot(df['UNRATE'], color='#2F6DB5')
ax.set_ylabel('Unemployment rate (%)')
plt.show()
```

Matplotlib also ships **style sheets** (QuantEcon 11.3.5). `plt.style.use('ggplot')` changes every chart after it, for the rest of the session, so switch back with `plt.style.use('default')` when you are done.

```python
print(plt.style.available)
```

```python
plt.style.use('ggplot')

fig, ax = plt.subplots()
ax.plot(df['CHG12'])
ax.set_title('Same data, ggplot style')
plt.show()

plt.style.use('default')   # back to normal for the rest of the notebook
```

Individual settings live in `plt.rcParams`, a dictionary of matplotlib's defaults. Changing one changes every later chart, so set them at the top of a notebook, and use `plt.rcdefaults()` to undo.

```python
plt.rcParams['figure.figsize'] = (8, 4)
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.spines.right'] = False

fig, ax = plt.subplots()
ax.plot(df['UNRATE'])
plt.show()

plt.rcdefaults()   # back to matplotlib's defaults
```

### Saving

`fig.savefig(filename)` writes the figure to disk; the extension picks the format (Sheppard 15.6).

- `.png` for slides and the web. Use `dpi=200` or more so it is sharp.
- `.pdf` or `.svg` for papers. These are vector files and stay sharp at any size.
- `bbox_inches='tight'` trims the empty border.

With the pyplot style, call `plt.savefig` **before** `plt.show()`, or you save a blank image. `fig.savefig` works any time, because you still hold `fig`.

```python
fig, ax = econ_axes()
ax.plot(df['UNRATE'], color='#2F6DB5')
ax.set_ylabel('Unemployment rate (%)')

fig.savefig('unrate.png', dpi=200, bbox_inches='tight')
fig.savefig('unrate.pdf', bbox_inches='tight')
plt.close(fig)

print('saved unrate.png and unrate.pdf next to this notebook')
```

## Optional, at home: math in labels, and 3D

Matplotlib understands LaTeX-style math between `$` signs. Use a raw string, `r'...'`, so Python leaves the backslashes alone (Sheppard 15.3.6, QuantEcon 11.2.3).

```python
rng = np.random.default_rng(630)
p = 50 * np.exp(np.cumsum(0.0004 + 0.01 * rng.standard_normal(250)))

fig, ax = plt.subplots()
ax.plot(p)
ax.set_xlabel(r'time ($\tau$)')
ax.set_ylabel('Price')
ax.set_title(r'Geometric random walk: $d \ln p_t = \mu\,dt + \sigma\,dW_t$')
plt.show()
```

3D plots are occasionally useful for a function of two variables, such as a bivariate density or a production function. Sheppard's verdict: 3D graphics are "rarely necessary, and often distracting". A contour plot usually shows the same thing more clearly.

```python
def cobb_douglas(k, l, alpha=0.3):
    return k**alpha * l**(1 - alpha)

k, l = np.meshgrid(np.linspace(0.1, 5, 60), np.linspace(0.1, 5, 60))
output = cobb_douglas(k, l)

fig = plt.figure(figsize=(11, 4.5))
ax3d = fig.add_subplot(1, 2, 1, projection='3d')
ax3d.plot_surface(k, l, output, cmap='viridis')
ax3d.set(xlabel='Capital', ylabel='Labor', zlabel='Output')

ax2d = fig.add_subplot(1, 2, 2)
lines = ax2d.contour(k, l, output, levels=[1, 2, 3, 4], cmap='viridis')
ax2d.clabel(lines, fontsize=8)
ax2d.set(xlabel='Capital', ylabel='Labor', title='Isoquants')
fig.tight_layout()
plt.show()
```

## Summary

| You want to show | Use | Key arguments |
|---|---|---|
| change over time | `ax.plot(x, y)` | `color`, `linestyle`, `linewidth`, `marker`, `label` |
| a relationship | `ax.scatter(x, y)` | `s`, `c`, `alpha`, `cmap` |
| a comparison across categories | `ax.bar`, `ax.barh` | `color`, `width` / `height` |
| a distribution | `ax.hist(x)` | `bins`, `density` |
| parts of a whole | `ax.pie(x)` | `labels`, `autopct` (usually a bar is better) |
| periods such as recessions | `ax.fill_between(x, y1, y2, where=...)` | `color`, `alpha` |
| a reference level | `ax.axhline(y)`, `ax.axvline(x)` | `linestyle` |

Every chart you hand to someone else needs: a title that states the finding, axis labels with units, a legend if there is more than one series, a source note, and a check that the data supports the title.

Errors you will hit:

| Symptom | Fix |
|---|---|
| saved PNG is blank | call `savefig` before `plt.show()`, or use `fig.savefig` |
| `AttributeError: 'numpy.ndarray' object has no attribute 'plot'` | `plt.subplots(2, 2)` returns an array of axes. Use `axes[0, 0].plot` |
| `AttributeError: 'Axes' object has no attribute 'xlabel'` | on an axes it is `ax.set_xlabel` (same for `set_title`, `set_ylim`) |
| a legend with no entries | give each `plot` call a `label=` |
| dates on the x-axis overlap | `fig.autofmt_xdate()`, or a `mdates` locator |
| the line has a gap | there is a `NaN` in the data. Find it with `isnull()` before deciding what to do |
